```python
#Length of longest word

def longest_word(words):
    return max(len(w) for w in words)

s = input("Enter words: ").split()
print("Length of longest word:", longest_word(s))
```

```python
#pattern using nested loop
def pattern(n):
    for i in range (1,n+1):
        for j in range(i):
            print("*",end=" ")
        print()
    for i in range(n-1,0,-1):
        for j in range(i):
            print("*",end=" ")
        print()
n=int(input("Enter the range:"))
pattern(n)
```

```python
#factors of a number
def factors(n):
    for i in range(1, n + 1):
        if n % i == 0:
            print(i, end=" ")

n = int(input("Enter a number: "))
factors(n)
```

```python
#area of square,rectangle and triangle using lambda functions
square=lambda a: a*a
rectangle=lambda l,b: l*b
triangle=lambda b,h: 0.5*b*h
a=float(input("Enter side of square:"))
print("Area of square=",square(a))
l=float(input("Enter length of rectangle:"))
b=float(input("Enter breadth of rectangle:"))
print("Area of rectangle=",rectangle(l,b))
b=float(input("Enter base of triangle:"))
h=float(input("Enter height of triangle:"))
print("Area of triangle=",triangle(b,h))
```
