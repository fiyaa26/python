```python
#change the string
s=input(" enter the string")
new_string=s[-1] + s[0:-1]+s[1]
print("new_string = ",new_string)
```

```python
# first 2 and last 2 
s=input("enter the string")
new_string=s[:2]+s[-2:] 
print("new string =",new_string)
```
