```python
#Count the occurences of each word in a line of text.
text= input("Enter a line of text:")
words= text.split()
count={}
for word in words:
    if word in count:
        count[word]=count[word]+1
    else:
        count[word]=1
print("Occurence")
for word in count:
    print(word,":",count[word])
```

```python
#Prompt the user for a list of integers .
#For all values greater than 100,store 'over' instead
num=list(map(int,input("Enter list of numbers seperated by space:").split()))
for i in range(len(num)):
    if num[i] > 100:
        num[i]= 'over'
print("List :",num)
```

```python
#Store a list of first names. Count the occurrences 'a' within a list
names=input("Enter names seperated by space:").split()
count=0
for name in names:
    for x in name:
        if x =='a' or x =='A':
            count=count+1
print("Number of A or a:",count)
```

```python
#Find gcd of 2 numbers
a=int(input("Enter the  first number :"))
b=int(input("Enter the  second number :"))
gcd=1
for i in range(1,min(a,b)+1):
    if a % i==0 and b%i==0:
        gcd = i
print("GCD:",gcd)
```

```python
#From a list of integers, create a list removing even numbers
numbers = list(map(int,input("Enter numbers separated by spaces: ").split()))
new_list = []
for num in numbers:
    if num % 2 != 0:
        new_list.append(num)
print("List after removing even numbers:", new_list)
```

```python
#Program to find factorial of a number
n = int(input("Enter a number: "))
fact = 1
for i in range(1, n + 1):
    fact = fact * i
print("Factorial =", fact)
```

```python
#Generate Fibanocci series of N terms
n = int(input("Enter the number of terms: "))
a = 0
b = 1
print("Fibonacci series:")
for i in range(n):
    print(a, end=" ")
    c = a + b
    a = b
    b = c
```

```python
#Generate a list of four digit numbers in a given range with all their digits even and the number is a perfect square
import math
start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))
print("Perfect square Numbers:")
for n in range(start, end + 1):
    if 1000 <= n <= 9999:
        root = math.isqrt(n)
        if root * root == n:
            temp = n
            all_even = True
            while temp > 0:
                digit = temp % 10
                if digit % 2 != 0:
                    all_even = False
                    break
                temp = temp // 10
            if all_even:
                print(n)
```

```python
#Enter 2 lists of integers.
#Check(a)whether list are of same length(b) whether list sums to same value(c) whether any value occur in both.
list1=list(map(int,input("Enter list 1:").split()))
list2=list(map(int,input("Enter list 2:").split()))
if len(list1) == len(list2):
    print("Same length")
else:
    print("Not same length")
if sum(list1) == sum(list2):
    print("Same sum")
else:
    print("Not same sum")
common=False
for n in list1:
    if n in list2:
        common=True
        break
if common:
    print("Common elements occured")
else:
    print("No Common elements occured")
```

```python
#Display the given pyramid with step number accepted from user
n = int(input("Enter the number of rows: "))
print("Pyramid:")
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(i * j, end="  ")
    print()
```
