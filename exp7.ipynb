```python
#Write a python program to printout all colors from colorset 1 not contained in colorset 2.
set1= set(input("Enter colors for set 1(seperate by comma):").split(","))
set2= set(input("Enter colors for set 2(seperate by comma):").split(","))
print("Colorset 1 :",set1)
print("Colorset 2 :",set2)
result= set1 - set2
print("Colors in set 1 not in set 2 :",result)
```

```python
#Write a python program to printout all colors from colorset 1 and colorset 2.
set1= set(input("Enter colors for set 1(seperate by comma):").split(","))
set2= set(input("Enter colors for set 2(seperate by comma):").split(","))
print("Colorset 1 :",set1)
print("Colorset 2 :",set2)
result= set1 | set2
print("Colors in both set :",result)
```
