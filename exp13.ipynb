```python
%%writefile rectangle.py
def area(l, b):
    return round(l * b, 2)

def perimeter(l, b):
    return round(2 * (l + b), 2)
```

```python
%%writefile circle.py
import math

def area(r):
    return round(math.pi * r * r, 2)

def perimeter(r):
    return round(2 * math.pi * r, 2)
```

```python
import rectangle
import circle

while True:
    print("\n----- MENU -----")
    print("1. Rectangle")
    print("2. Circle")
    print("3. Exit")

    ch = int(input("Enter your choice: "))

    if ch == 1:
        l = float(input("Enter length: "))
        b = float(input("Enter breadth: "))

        print("Area of Rectangle =", rectangle.area(l, b))
        print("Perimeter of Rectangle =", rectangle.perimeter(l, b))

    elif ch == 2:
        r = float(input("Enter radius: "))

        print("Area of Circle =", circle.area(r))
        print("Perimeter of Circle =", circle.perimeter(r))

    elif ch == 3:
        print("Exiting Program...")
        break

    else:
        print("Invalid Choice")
```
