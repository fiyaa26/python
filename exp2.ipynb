```python
#Python program to convert Celsius to fahrenheit (T(F)=T(c)*(9/5)+32)
n= float(input("Enter the temperature in celcius :"))
f=n*(9/5)+32
print("Temperature in fahrenheit : ",round(f,1))
```

```python
#Python program to convert kilometers to miles(1 km = 0.62137)
km= int(input("Enter the distance in kilometer: "))
n=0.62137
mi=km*n
print("Distance in Miles :",round(mi,2))
```
